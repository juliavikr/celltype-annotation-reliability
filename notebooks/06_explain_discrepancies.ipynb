# 06 · Explaining the discrepancies (Finding 4)
Arianna's question: can disagreements be explained by **annotation strategy** or **ontology usage**?
1. Join `results/agreement.csv` with `data/processed/annotation_strategies.csv`. Do same-strategy pairs agree more?
2. For datasets with author labels: compare each author label with its CL term. When was the label made broader or changed in meaning?

```python
import pandas as pd
agreement = pd.read_csv("../results/agreement.csv")
strat = pd.read_csv("../data/processed/annotation_strategies.csv")
s = strat.set_index("dataset_id").strategy
agreement["same_strategy"] = agreement.train.map(s) == agreement.test.map(s)
agreement.groupby("same_strategy")[["exact", "compatible", "conflict"]].mean()
```
