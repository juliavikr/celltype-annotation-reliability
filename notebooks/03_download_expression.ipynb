# 03 · Expression data for the chosen datasets
Two options:
- **Census slice** (`get_anndata`): standardized, only curated labels.
- **Original dataset file** (`download_source_h5ad`): bigger, but keeps the authors' own label columns. You need these for Finding 4 (did ontology mapping change the label?).

```python
import cellxgene_census
CHOSEN = []
census = cellxgene_census.open_soma(census_version="stable")
for ds in CHOSEN:
    adata = cellxgene_census.get_anndata(census, organism="Homo sapiens",
                                         obs_value_filter=f"dataset_id == '{ds}' and is_primary_data == True")
    # subsample if memory is tight: sc.pp.subsample(adata, n_obs=20000)
    adata.write_h5ad(f"../data/processed/{ds}.h5ad")
census.close()
```
